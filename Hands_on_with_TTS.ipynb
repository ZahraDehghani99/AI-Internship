````python
from transformers import VitsModel, AutoTokenizer
import torch

model = VitsModel.from_pretrained("facebook/mms-tts-fas")
tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-fas")

text = "سلام خوبی؟ این یه تست از ماژول تبدیل متن به گفتار است"
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    output = model(**inputs).waveform
````

````python
from IPython.display import Audio

Audio(output, rate=model.config.sampling_rate)
````

````python
import torchaudio
import torch

from transformers import AutoProcessor, SeamlessM4TModel
processor = AutoProcessor.from_pretrained("facebook/hf-seamless-m4t-medium")
model = SeamlessM4TModel.from_pretrained("facebook/hf-seamless-m4t-medium")
````

````python
# Process some input text as well:
text_inputs = processor(text = "سلام خوبی؟ این یه تست از ماژول تبدیل متن به گفتار است", src_lang="pes", return_tensors="pt")
````

````python
audio_array_from_text = model.generate(**text_inputs, tgt_lang="pes")[0].cpu().numpy().squeeze()
````

````python
# Save audio
torchaudio.save("output.wav", torch.tensor(audio_array_from_text).unsqueeze(0), 16000)
````

````python
TEXT="""
برای رفع مشکل عدم اجرا شدن سرویس الستیک (Elasticsearch) در محصول APKGatev5، مراحل زیر را به ترتیب انجام دهید:

1. **بررسی دسترسی‌های فایل سرویس الستیک**
   اگر سرویس الستیک شروع به کار نمی‌کند، ابتدا سطح دسترسی فایل `/usr/local/etc/rc.d/elasticsearch` را چک کنید. با استفاده از دستور `chmod` دسترسی‌های `read` و `write` را به سرویس اعطا کنید:
   ```bash
   chmod +x /usr/local/etc/rc.d/elasticsearch
   ```
   این دستور اطمینان می‌دهد که سرویس مجاز به اجرای فایل است [1].

2. **بررسی لاگ‌های سرویس الستیک**
   اگر سرویس هنوز اجرا نشود، لاگ‌های سرویس را در مسیر `/var/log/elasticsearch/elasticsearch.log` بررسی کنید. اگر خطایی مانند `java.lang.OutOfMemoryError: Java heap space` مشاهده شد، مقدار حافظه اختصاص داده شده به الستیک را افزایش دهید. برای این کار به فایل `/usr/local/etc/rc.d/elasticsearch` مراجعه کرده و پارامترهای زیر را ویرایش کنید:
   ```bash
   ES_JAVA_OPTS="
   -Xms2g \
   -Xmx2g \
   "
   ```
   مقدار `2g` می‌تواند به سایر مقادیر (مثلاً `4g`) تغییر کند [6].

3. **ریستارت سرویس الستیک**
   پس از تغییرات، سرویس الستیک را با دستور زیر ریستارت کنید:
   ```bash
   /usr/local/etc/rc.d/elasticsearch onerestart
   ```
   این دستور سرویس را با تنظیمات جدید شروع می‌کند [2].

4. **حذف ایندکس‌های قدیمی**
   اگر مشکل ادامه یابد، ممکن است ایندکس‌های قدیمی در دیتابیس الستیک باعث مشکل شده باشند. ابتدا لیست ایندکس‌ها را با دستور زیر مشاهده کنید:
   ```bash
   curl -XGET "localhost:9200/_cat/indices"
   ```
   سپس ایندکس‌های مورد نظر (مثلاً مربوط به ماه خاصی) را حذف کنید:
   ```bash
   curl -XDELETE "localhost:9200/*2023.05*"
   ```
   پس از حذف، سرویس الستیک را دوباره ریستارت کنید [2].

5. **بررسی حافظه سیستم**
   اگر سرویس هنوز اجرا نشود، ممکن است مشکل از کمبود حافظه (RAM) باشد. با دستور زیر وضعیت حافظه سیستم را بررسی کنید:
   ```bash
   vmstat -s
   ```
   اگر حافظه کافی نباشد، سرویس‌های دیگر مانند `zeek` را متوقف کنید تا فضای حافظه آزاد شود:
   ```bash
   /usr/local/etc/rc.d/zeek onestop
   /usr/local/etc/rc.d/elasticsearch onerestart
   ```
   این راهکار موقتی است و پس از رفع مشکل، سرویس‌ها را دوباره فعال کنید [2].

6. **نصب مجدد الستیک**
   اگر هیچ‌یک از راهکارهای بالا کار نکرد، ممکن است نسخه نصبی الستیک آسیب دیده باشد. برای رفع این مشکل:
   - الستیک را حذف کنید:
     ```bash
     pkg remove elasticsearch5-5.6.8_5
     ```
   - فایل نصبی جدید را در مسیر `root` قرار دهید و نصب کنید:
     ```bash
     pkg add elasticsearch5-5.6.8_5
     pkg install elasticsearch5-5.6.8_5
     ```
   - سرویس الستیک و `fluentd` را ریستارت کنید:
     ```bash
     /usr/local/etc/rc.d/elasticsearch onerestart
     /usr/local/etc/rc.d/fluentd onerestart
     ```
   این روش مخصوصاً در صورتی موثر است که خطایی در فایل‌های نصب وجود داشته باشد [3].

**نکته مهم:**
اگر پس از انجام تمام مراحل مشکل همچنان پابرجا باشد، ممکن است نیاز به بررسی جزئیات بیشتری از سیستم داشته باشید. در این صورت، با تیم پشتیبانی شرکت APK تماس بگیرید.

مراجع:
[1]: https://confluence.apk-group.net/pages/viewpage.action?pageId=78776060
[2]: https://confluence.apk-group.net/display/AP/Elasticsearch+service
[3]: https://confluence.apk-group.net/pages/viewpage.action?pageId=78776063
[6]: https://confluence.apk-group.net/pages/viewpage.action?pageId=78776057

"""

with open("input.txt", "w", encoding="utf-8") as f:
    f.write(TEXT)

````
